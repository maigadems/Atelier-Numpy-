## Partie 1 : Listes Python vs tableaux NumPy

Températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0, 25.8

1. Liste Python
2. Tableau NumPy équivalent
3. Multiplication par 2 sur les deux
4. Différence observée ?

````python
import numpy as np

temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
temperatures_liste
````

````python
temperatures_array = np.array(temperatures_liste)
temperatures_array
````

````python
temperatures_liste * 2
````

````python
temperatures_array * 2
````

**Différence observée** :

- `liste * 2` **duplique** la liste (concatène la liste avec elle-même) → on obtient 10 éléments.
- `array * 2` applique la multiplication **élément par élément** (opération vectorisée) → on obtient toujours 5 éléments, chacun multiplié par 2.

C'est la différence fondamentale entre une liste Python (conteneur générique) et un tableau NumPy (structure numérique optimisée pour le calcul).

On réutilise `temperatures_jour` (24 valeurs, l'indice `i` correspond à l'heure `i`h,

Ce qu'il faut retenir :

- **Indexation simple** : `tableau[i]` accède à l'élément d'indice `i` — comme pour une liste
  Python, le premier élément est à l'indice `0` (donc `temperatures_jour[8]` correspond bien
  à 8h, pas à la 8ème mesure « humaine »).
- **Indice négatif** : `tableau[-1]` accède au dernier élément en partant de la fin, sans avoir
  à connaître la longueur du tableau. `-1` = dernier, `-2` = avant-dernier, etc.
- **Slicing `[début:fin]`** : renvoie une *sous-partie* du tableau, de l'indice `début` inclus
  à l'indice `fin` **exclu**. C'est pour ça qu'on écrit `[8:13]` (et pas `[8:12]`) pour obtenir
  les heures 8, 9, 10, 11 **et** 12.
- **Slicing avec pas `[début:fin:pas]`** : `[:6]` veut dire « du début jusqu'à l'indice 6 exclu »
  (donc les indices 0 à 5, soit les 6 premières heures). `[::2]` veut dire « du début à la fin,
  en avançant de 2 en 2 » — ici, comme les indices pairs correspondent aux heures paires, ça
  donne directement les températures de 0h, 2h, 4h, ..., 22h.
- Toutes ces opérations renvoient un **nouveau tableau** (sauf l'indexation simple qui renvoie
  un scalaire), et surtout : sur un tableau NumPy, le slicing renvoie une **vue** (pas une copie)
  qui partage la même mémoire que le tableau d'origine — modifier la vue modifierait aussi
  `temperatures_jour`. C'est une différence importante avec les listes Python, où trancher crée
  toujours une copie.

## Partie 5 : Filtrage booléen

Toujours sur `temperatures_jour`, on cherche à détecter des situations inhabituelles.

1. Températures supérieures à 30 °C
2. Températures « normales » (entre 20 °C et 30 °C)
3. `anomalies` : tableau de booléens, `True` si la température dépasse 35 °C
4. Nombre de températures anormales, à partir de `anomalies`

````python
temperatures_jour[temperatures_jour > 30]
````

````python
temperatures_jour[(temperatures_jour >= 20) & (temperatures_jour <= 30)]
````

````python
anomalies = temperatures_jour > 35
anomalies
````

````python
anomalies.sum()
````

Le mécanisme central ici s'appelle le **masque booléen** (*boolean masking*) :

- `temperatures_jour > 30` ne renvoie **pas** un seul `True`/`False`, mais un tableau de
  24 booléens (un par élément), résultat de la comparaison élément par élément — encore
  la vectorisation vue en Partie 1.
- Utiliser ce tableau de booléens comme indice — `temperatures_jour[masque]` — ne garde
  que les éléments où le masque vaut `True`. C'est ce qu'on appelle le *filtrage booléen*.
- Pour combiner deux conditions, on utilise `&` (et) / `|` (ou) sur les tableaux, **pas**
  les mots-clés Python `and` / `or` qui ne fonctionnent que sur une seule valeur booléenne
  à la fois, pas sur un tableau entier. Et il faut **parenthéser chaque condition**
  (`(... >= 20) & (... <= 30)`), sinon Python évalue `&` avant `>=`/`<=` à cause des
  priorités des opérateurs, ce qui casse l'expression.
- `anomalies = temperatures_jour > 35` est lui-même un tableau de booléens de même taille
  que `temperatures_jour` — pas les valeurs anormales elles-mêmes, mais un « marqueur »
  `True`/`False` pour chaque heure.
- `anomalies.sum()` fonctionne parce qu'en Python, `True` vaut `1` et `False` vaut `0` :
  additionner un tableau de booléens revient donc à **compter les `True`**. Ici, le résultat
  est `0` : avec la graine `42` utilisée en Partie 3, aucune des 24 températures ne dépasse
  35 °C (la plus haute valeur observée était d'environ 33.4 °C) — c'est une journée sans
  anomalie, et c'est un résultat parfaitement valide.

## Partie 6 : Manipulation des dimensions

On suppose 5 capteurs (dans le 1er bâtiment).

1. Matrice aléatoire des températures des 5 capteurs sur une journée (24h)

7. Mesures des capteurs 2 à 4 entre 8h et 12h

````python
temperatures_capteurs = rng.normal(loc=25, scale=6, size=(5, 24))
temperatures_capteurs.shape
````

````python
temperatures_capteurs[0]
````

````python
temperatures_capteurs[4]
````

````python
temperatures_capteurs[:, 10]
````

````python
temperatures_capteurs[1:4, 8:13]
````

Ce qu'il faut retenir :

- **`size=(5, 24)`** : on demande directement un tableau à 2 dimensions — 5 lignes
  (les capteurs), 24 colonnes (les heures). `shape` confirme `(5, 24)`.
- **`flatten()`** aplati la matrice en un vecteur 1D de 5×24 = 120 valeurs, en parcourant
  ligne par ligne (`row-major`, ordre `'C'` par défaut). `flatten()` renvoie toujours une
  **copie** ; sa cousine `ravel()` fait la même chose mais renvoie une **vue** quand c'est
  possible (donc plus rapide, mais modifier le résultat peut modifier l'original — à utiliser
  avec prudence).
- **`.T`** (transposée) échange les axes : une matrice `(5, 24)` devient `(24, 5)`. Les lignes
  deviennent colonnes et inversement — pratique par exemple pour passer d'une vue
  "un capteur par ligne" à "une heure par ligne".
- **Indexation 2D** : `matrice[i]` (ou `matrice[i, :]`) sélectionne la **ligne** `i`, donc ici
  toutes les mesures d'un capteur. Le capteur 1 (au sens humain, "premier capteur") correspond
  à l'indice `0` ; le capteur 5 à l'indice `4` — toujours ce décalage entre numérotation humaine
  (1 à 5) et indices Python (0 à 4).
- **`matrice[:, 10]`** : `:` sur le premier axe = « toutes les lignes », `10` sur le second axe
  = « seulement la colonne 10 ». Résultat : la mesure de 10h pour chacun des 5 capteurs.
- **`matrice[1:4, 8:13]`** combine un slicing sur chaque axe : `1:4` sélectionne les capteurs
  d'indices 1, 2, 3 (donc les capteurs "humains" 2, 3, 4), et `8:13` les heures 8h à 12h
  incluses (même piège que Partie 4 : borne de fin exclue, donc `13` pour aller jusqu'à 12h).
  Résultat : une sous-matrice `(3, 5)`.

## Partie 7 : Concaténation

On suppose un 2nd bâtiment avec 3 capteurs.

1. Matrice aléatoire des températures des 3 capteurs sur une journée
2. Fusionner, de deux manières différentes, les mesures des 8 capteurs (5 + 3) sur une journée

````python
temperatures_batiment2 = rng.normal(loc=25, scale=6, size=(3, 24))
temperatures_batiment2.shape
````

````python
temperatures_8_capteurs_v1 = np.concatenate([temperatures_capteurs, temperatures_batiment2], axis=0)
temperatures_8_capteurs_v1.shape
````

````python
temperatures_8_capteurs_v2 = np.vstack([temperatures_capteurs, temperatures_batiment2])
temperatures_8_capteurs_v2.shape
````

````python
np.array_equal(temperatures_8_capteurs_v1, temperatures_8_capteurs_v2)
````

- **`np.concatenate([a, b], axis=0)`** : la fonction générique pour empiler des tableaux.
  `axis=0` veut dire qu'on empile **le long des lignes** — donc on ajoute des lignes (des
  capteurs en plus), en gardant le même nombre de colonnes (24 heures). Il faut que toutes
  les dimensions *sauf* celle sur laquelle on concatène soient identiques (ici, les deux
  matrices ont bien 24 colonnes).
- **`np.vstack([a, b])`** ("vertical stack") fait exactement la même chose que
  `concatenate(..., axis=0)`, mais avec un nom plus explicite pour le cas 2D. C'est un
  raccourci pratique — `np.array_equal(...)` confirme que les deux résultats sont identiques.
- Il existe le pendant `np.hstack` (empilage horizontal, équivalent à `axis=1`), qu'on
  utiliserait si on voulait ajouter des heures plutôt que des capteurs.

## Partie 8 : Calcul scientifique

On revient sur `temperatures_jour` (les 24 températures du 1er bâtiment, Partie 3).

1. Minimum, maximum, somme, moyenne, médiane, variance, écart-type
2. Standardiser puis vérifier moyenne et écart-type du résultat
3. Normaliser (Min-Max) puis vérifier le résultat
4. Convertir en Fahrenheit

````python
print("min     :", temperatures_jour.min())
print("max     :", temperatures_jour.max())
print("somme   :", temperatures_jour.sum())
print("moyenne :", temperatures_jour.mean())
print("médiane :", np.median(temperatures_jour))
print("variance:", temperatures_jour.var())
print("écart-type:", temperatures_jour.std())
````

````python
temperatures_standardisees = (temperatures_jour - temperatures_jour.mean()) / temperatures_jour.std()
print("moyenne :", temperatures_standardisees.mean())
print("écart-type:", temperatures_standardisees.std())
````

````python
temperatures_normalisees = (temperatures_jour - temperatures_jour.min()) / (temperatures_jour.max() - temperatures_jour.min())
print("min:", temperatures_normalisees.min())
print("max:", temperatures_normalisees.max())
````

````python
temperatures_fahrenheit = temperatures_jour * 9 / 5 + 32
temperatures_fahrenheit
````

À retenir :

- Toutes ces statistiques (`min`, `max`, `sum`, `mean`, `var`, `std`) existent en deux formes
  équivalentes : méthode sur le tableau (`temperatures_jour.mean()`) ou fonction NumPy
  (`np.mean(temperatures_jour)`). `median` n'existe **que** comme fonction (`np.median(...)`),
  pas comme méthode — une petite incohérence de l'API à connaître.
- **Variance** (`var`) mesure la dispersion moyenne au carré autour de la moyenne ;
  **écart-type** (`std`) en est la racine carrée, donc exprimé dans la même unité que les
  données (°C) — plus interprétable.
- **Standardisation** (ou *z-score*) : `(x - moyenne) / écart-type`. Par construction
  mathématique, le résultat a **toujours** une moyenne ≈ 0 et un écart-type ≈ 1 (les petits
  écarts vus ici viennent uniquement des arrondis flottants). C'est le prétraitement classique
  avant d'envoyer des données à un modèle de Machine Learning, pour que toutes les variables
  (température, humidité, pression...) soient sur une échelle comparable.
- **Normalisation Min-Max** : `(x - min) / (max - min)` ramène toutes les valeurs dans
  l'intervalle `[0, 1]` — le minimum du tableau devient exactement `0`, le maximum exactement
  `1`. Utile pour des algorithmes sensibles à l'échelle des données (réseaux de neurones,
  k-NN...), mais contrairement à la standardisation, très sensible aux valeurs extrêmes
  (une seule valeur aberrante écrase toute l'échelle).
- **Conversion Fahrenheit** (`x * 9/5 + 32`) : encore une opération vectorisée qui s'applique à
  chaque élément d'un coup, sans boucle — le fil rouge de tout l'atelier depuis la Partie 1.

## Partie 9 : Broadcasting

1. `temperatures_vsd` : températures mesurées vendredi, samedi et dimanche, dans le 2nd
   bâtiment (3 capteurs) — une ligne par jour, une colonne par capteur, donc `shape = (3, 3)`
2. `correction = np.array([1, -0.5, 0.2])` : une valeur de calibration par capteur.
   Appliquer automatiquement la correction : `mesures_corrigees = temperatures_vsd + correction`

````python
temperatures_vsd = rng.normal(loc=25, scale=6, size=(3, 3))  # 3 jours x 3 capteurs
temperatures_vsd
````

````python
correction = np.array([1, -0.5, 0.2])
mesures_corrigees = temperatures_vsd + correction
mesures_corrigees
````

`temperatures_vsd` a la forme `(3, 3)` (3 jours × 3 capteurs) et `correction` a la forme
`(3,)`. Ce sont des formes différentes, et pourtant `+` fonctionne directement : c'est le
**broadcasting**.

La règle (NumPy compare les formes de droite à gauche) :

```
temperatures_vsd :  (3, 3)
correction       :     (3,)
```

- NumPy aligne les dimensions en partant de la **droite** : `3` (capteurs de `temperatures_vsd`)
  et `3` (`correction`) correspondent → ok.
- Il manque une dimension à `correction` à gauche : NumPy la traite comme si elle valait
  implicitement `1`, puis « l'étire » virtuellement pour matcher les 3 jours — **sans dupliquer
  physiquement les données en mémoire**, uniquement pendant le calcul.
- Concrètement, `correction` est appliquée **à l'identique sur chaque ligne** (chaque jour) :
  la même correction par capteur, tous les jours.

Règle générale du broadcasting entre deux tableaux : en comparant les tailles axe par axe
depuis la droite, chaque paire de dimensions doit être **soit égale, soit valoir 1** (auquel
cas elle est étirée). Si les dimensions ne correspondent pas et qu'aucune ne vaut 1,
NumPy lève une erreur (`ValueError: operands could not be broadcast together`).

C'est ce même mécanisme, en plus simple, qu'on utilisait déjà depuis la Partie 1 :
`temperatures_array * 2` fonctionne parce que le scalaire `2` (considéré comme de forme `()`)
est broadcastable contre n'importe quelle forme.
