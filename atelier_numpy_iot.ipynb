## Partie 1 : Listes Python vs tableaux NumPy

Températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0, 25.8

1. Liste Python
2. Tableau NumPy équivalent
3. Multiplication par 2 sur les deux
4. Différence observée ?

```python
import numpy as np

temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
temperatures_liste
```

```python
temperatures_array = np.array(temperatures_liste)
temperatures_array
```

```python
temperatures_liste * 2
```

```python
temperatures_array * 2
```

**Différence observée** :

- `liste * 2` **duplique** la liste (concatène la liste avec elle-même) → on obtient 10 éléments.
- `array * 2` applique la multiplication **élément par élément** (opération vectorisée) → on obtient toujours 5 éléments, chacun multiplié par 2.

C'est la différence fondamentale entre une liste Python (conteneur générique) et un tableau NumPy (structure numérique optimisée pour le calcul).

les mêmes résultats en relançant le notebook.

```python
rng = np.random.default_rng(42)
mesures_aleatoires = rng.normal(loc=25, scale=3, size=100)
mesures_aleatoires[:10]
```
