## Partie 1 : Listes Python vs tableaux NumPy

Températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0, 25.8

1. Liste Python
2. Tableau NumPy équivalent
3. Multiplication par 2 sur les deux
4. Différence observée ?

```python
import numpy as np

temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
temperatures_liste
```

```python
temperatures_array = np.array(temperatures_liste)
temperatures_array
```

```python
temperatures_liste * 2
```

```python
temperatures_array * 2
```

**Différence observée** :

- `liste * 2` **duplique** la liste (concatène la liste avec elle-même) → on obtient 10 éléments.
- `array * 2` applique la multiplication **élément par élément** (opération vectorisée) → on obtient toujours 5 éléments, chacun multiplié par 2.

C'est la différence fondamentale entre une liste Python (conteneur générique) et un tableau NumPy (structure numérique optimisée pour le calcul).

On garde ce tableau `temperatures_jour` : il sera réutilisé dans les Parties 4 et 5.

```python
temperatures_jour.shape
```

- `ndim` : nombre d'axes/dimensions du tableau (ici 1, c'est un vecteur — on aura des matrices à `ndim=2` en Partie 6).
- `size` : nombre total de valeurs stockées (24, indépendamment du nombre de dimensions).
- `dtype` : type des éléments. NumPy stocke **un seul type homogène** par tableau (contrairement à une liste Python qui peut mélanger les types) — ici `float64`, un flottant sur 64 bits. C'est ce typage homogène qui permet à NumPy d'être rapide : il sait à l'avance combien d'octets occupe chaque élément.
- `shape` : un tuple donnant la taille de chaque dimension. Ici `(24,)` — un tuple à un seul élément car le tableau est 1D. En Partie 6, une matrice de 5 capteurs × 24 heures aura `shape = (5, 24)`.

## Partie 4 : Indexation et slicing

On réutilise `temperatures_jour` (24 valeurs, l'indice `i` correspond à l'heure `i`h,

Ce qu'il faut retenir :

- **Indexation simple** : `tableau[i]` accède à l'élément d'indice `i` — comme pour une liste
  Python, le premier élément est à l'indice `0` (donc `temperatures_jour[8]` correspond bien
  à 8h, pas à la 8ème mesure « humaine »).
- **Indice négatif** : `tableau[-1]` accède au dernier élément en partant de la fin, sans avoir
  à connaître la longueur du tableau. `-1` = dernier, `-2` = avant-dernier, etc.
- **Slicing `[début:fin]`** : renvoie une *sous-partie* du tableau, de l'indice `début` inclus
  à l'indice `fin` **exclu**. C'est pour ça qu'on écrit `[8:13]` (et pas `[8:12]`) pour obtenir
  les heures 8, 9, 10, 11 **et** 12.
- **Slicing avec pas `[début:fin:pas]`** : `[:6]` veut dire « du début jusqu'à l'indice 6 exclu »
  (donc les indices 0 à 5, soit les 6 premières heures). `[::2]` veut dire « du début à la fin,
  en avançant de 2 en 2 » — ici, comme les indices pairs correspondent aux heures paires, ça
  donne directement les températures de 0h, 2h, 4h, ..., 22h.
- Toutes ces opérations renvoient un **nouveau tableau** (sauf l'indexation simple qui renvoie
  un scalaire), et surtout : sur un tableau NumPy, le slicing renvoie une **vue** (pas une copie)
  qui partage la même mémoire que le tableau d'origine — modifier la vue modifierait aussi
  `temperatures_jour`. C'est une différence importante avec les listes Python, où trancher crée
  toujours une copie.
